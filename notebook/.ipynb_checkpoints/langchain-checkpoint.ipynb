````python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
````

````python
llm = ChatOllama(
    model="qwen2.5:14b",
    base_url="http://host.docker.internal:11434",   # 关键：访问主机
    temperature=0.2,
    num_ctx=8192,          # Agent 建议大一点的上下文
    num_gpu=999,
)
````

````python
from typing import TypedDict, Annotated, Literal


class AgentState(TypedDict):
    message: Annotated[list, "聊天历史消息"]
    next: str
    resume_context: dict
    task_plan: list
````

````python
from langchain_core.tools import tool
@tool
def get_current_date() -> str:
    """返回当前日期和时间，用于时间敏感的任务"""
    from datetime import datetime
    return datetime.now().strftime("%Y-%m-%d %H:%M:%S")

@tool
def calculate(expression: str) -> str:
    """安全地执行简单的数学计算"""
    try:
        # 限制只允许基础运算，生产环境建议用更安全的 sandbox
        allowed_names = {"__builtins__": {}}
        result = eval(expression, allowed_names)
        return str(result)
    except Exception as e:
        return f"计算错误: {str(e)}"
tools = [get_current_date, calculate]
````

````python
from langchain.agents import create_agent
text = "你好，我是 Dev Agent"
style = f""" American English\
hip-hop street style
"""
prompt = f""" translate the text in English\
that is delimited by triple backticks\
into a style {style}
text:```{text}```
"""

agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=prompt
)
````

````python
from langchain_core.messages import AIMessage, HumanMessage


messages = [HumanMessage(content="今天是什么日期？请帮我计算 9876 * 543 的结果。")]

result = agent.invoke({"messages": messages})
messages = result["messages"]

# 找到最后一个 AIMessage
for msg in reversed(messages):
    if isinstance(msg, AIMessage) and msg.content:   # 确保有内容
        print("=== DevAgent 最终回答 ===")
        print(msg.content)
        break
````

````python
from pathlib import Path
resume_dir = Path("/workspace/data/resumes")

print("当前简历目录内容：")
for pdf_file in resume_dir.glob("*.pdf"):
    print(f"   📄 {pdf_file.name}")

if not any(resume_dir.glob("*.pdf")):
    print("\n⚠️  提示：请将你的简历PDF文件放入 devagent/data/resumes/ 文件夹中")
    print("   示例文件名：resume_sun.pdf")
````

````python
# ==================== 5. PDF 加载与文档切分 ====================
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

def load_and_split_resume(pdf_path: str):
    """加载并切分简历PDF"""
    loader = PyPDFLoader(pdf_path)
    documents = loader.load()
    
    print(f"✅ 成功加载 PDF，共 {len(documents)} 页")
    
    # 适合简历的文本切分器
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=800,
        chunk_overlap=100,
        separators=["\n\n", "\n", "。", "，", " ", ""],
        length_function=len,
    )
    
    splits = text_splitter.split_documents(documents)
    print(f"✅ 切分后得到 {len(splits)} 个文本块")
    
    return splits

# ==================== 执行加载 ====================
pdf_files = list(resume_dir.glob("*.pdf"))

if pdf_files:
    resume_path = pdf_files[0]   # 默认取第一个PDF
    print(f"正在处理简历: {resume_path.name}")
    
    resume_splits = load_and_split_resume(str(resume_path))
else:
    print("未找到PDF文件，请先放入简历后再运行此 cell")
    resume_splits = []
````

````python
# ==================== 6. 构建向量数据库 (Chroma) ====================
from langchain_chroma import Chroma
from langchain_ollama import OllamaEmbeddings


embeddings = OllamaEmbeddings(model="qwen2.5:14b", base_url="http://host.docker.internal:11434")
print(f"   模型: {llm.model}")
persist_directory = "/workspace/data/chroma_db"

# 如果已存在数据库，先删除旧的（开发阶段建议）
if Path(persist_directory).exists():
    import shutil
    shutil.rmtree(persist_directory)
    print("🗑️  已清理旧向量数据库")

vectorstore = Chroma.from_documents(
    documents=resume_splits,
    embedding=embeddings,
    persist_directory=persist_directory,
    collection_name="resume_knowledge"
)

print(f"✅ 向量数据库构建完成！共存储 {vectorstore._collection.count()} 条向量")
````

````python
# ==================== 7. 创建 RAG Retriever ====================

retriever = vectorstore.as_retriever(
    search_type="similarity", 
    search_kwargs={"k": 6}      # 每次检索返回6个最相关片段
)

print("✅ RAG Retriever 已创建")
````

````python
# ==================== 8. 测试 RAG 检索效果 ====================

test_queries = [
    "我的开发经验有多少年？主要技术栈是什么？",
    "我在 Citigroup 的工作内容是什么？",
    "我熟悉哪些 AI / LLM 相关技术？"
]

print("=== RAG 检索测试 ===\n")

for query in test_queries:
    print(f"🔍 查询: {query}")
    docs = retriever.invoke(query)
    
    print(f"   返回 {len(docs)} 个文档片段")
    for i, doc in enumerate(docs[:2]):   # 只显示前2条
        print(f"   [{i+1}] {doc.page_content[:150]}...")
    print("-" * 80)
````

````python
# ==================== 9. 封装 RAG Chain（为后续 Agent 准备） ====================

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

template = """你是一位专业的简历助手。请基于以下上下文信息，准确、专业地回答用户问题。

上下文：
{context}

问题：{question}

回答要求：
- 严格基于上下文信息回答，不要编造
- 保持专业语气
- 如果上下文无法回答，请明确说明

回答："""

prompt = ChatPromptTemplate.from_template(template)

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

rag_chain = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

print("✅ RAG Chain 封装完成，可直接用于 Agent")
````

````python
# ==================== 10. 测试完整 RAG Chain ====================

print("=== RAG Chain 端到端测试 ===\n")

# test_question = "请总结我的主要技术背景和优势"
test_question = "我能否胜任AI agent开发的工作？"

response = rag_chain.invoke(test_question)

print(response)
````

````python
import os
from dotenv import load_dotenv
load_dotenv(override=True)

print(os.getenv("TAVILY_API_KEY"))
@tool
def calculate(expression: str) -> str:
    """执行简单的数学计算"""
    try:
        allowed_names = {"__builtins__": {}}
        return str(eval(expression, allowed_names))
    except Exception as e:
        return f"计算错误: {str(e)}"
    

@tool
def search_resume(query: str) -> str:
    """使用 RAG 在用户简历中搜索相关信息"""
    docs = retriever.invoke(query)
    context = "\n\n".join([doc.page_content for doc in docs])
    return f"【简历检索结果】\n{context[:1500]}"   # 限制长度避免 token 爆炸

@tool
def web_search(query: str) -> str:
    """使用 Tavily 进行网页搜索（需要 TAVILY_API_KEY）"""
    try:
        from tavily import TavilyClient
        tavily = TavilyClient(api_key=os.getenv("TAVILY_API_KEY"))
        response = tavily.search(query=query, max_results=5)
        results = [f"标题: {r['title']}\n内容: {r['content'][:300]}" for r in response['results']]
        return "\n\n".join(results)
    except Exception as e:
        return f"网页搜索失败: {str(e)}（请确保设置了 TAVILY_API_KEY）"

tools = [get_current_date, calculate, search_resume, web_search]
print(f"✅ 已定义 {len(tools)} 个工具（包含 RAG 工具）")
````

````python
# ==================== 5. 使用 LangGraph 构建 ReAct Agent（推荐写法） ====================



# 专业的 System Prompt（突出你的背景）
from langchain_core.messages import SystemMessage


system_prompt = SystemMessage(content="""你是一位拥有10年开发经验的资深 AI Engineer，擅长 Python、C#/.NET、Transformer 和 LLM 系统架构。
你目前正在帮助用户基于他的个人简历进行职业发展支持，包括简历优化、面试准备、技术研究等。
回答时要专业、严谨、结构化。如果需要使用工具，请先思考再调用。""")

agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt,   # 正确的参数名（新版写法）
)

print("✅ LangGraph ReAct Agent 创建成功（支持多工具并行调用）")
````

- 定义清晰的 Agent 角色分工（Researcher、Coder、Reviewer、Orchestrator）
- 实现动态任务路由和协作
- 为后续产品级打磨打好基础

````python
# ==================== 2. 配置环境 & LLM ====================

import os
from dotenv import load_dotenv
from pathlib import Path

load_dotenv(override=True)

from langchain_ollama import ChatOllama
from langchain_ollama import OllamaEmbeddings

# === 速度优化后的 LLM 配置（根据你昨天反馈优化）===
llm = ChatOllama(
    model="qwen2.5:14b",
    base_url="http://host.docker.internal:11434",   # 关键：访问主机
    temperature=0.2,
    num_ctx=8192,          # Agent 建议大一点的上下文
    num_gpu=999
)

# embeddings = OllamaEmbeddings(model="nomic-embed-text")   # 使用轻量 embedding 模型加速
embeddings = OllamaEmbeddings(model="qwen2.5:14b", base_url="http://host.docker.internal:11434")

print(f"✅ LLM 初始化完成 → {llm.model} | Context: {llm.num_ctx}")
print("✅ Embedding 模型已切换为轻量版（速度更快）")
````

````python
# ==================== 3. 加载 RAG Retriever（复用 Day 2） ====================

from langchain_chroma import Chroma

vectorstore = Chroma(
    persist_directory="/workspace/data/chroma_db",
    embedding_function=embeddings,
    collection_name="resume_knowledge"
)

retriever = vectorstore.as_retriever(search_kwargs={"k": 6})

print(f"✅ RAG Retriever 加载成功，共有 {vectorstore._collection.count()} 条向量")
````

````python
# ==================== 4. 定义 Multi-Agent 专用 State ====================

from typing import TypedDict, Annotated, List, Sequence
from langchain_core.messages import BaseMessage
import operator

class AgentState(TypedDict):
    """Multi-Agent 系统核心状态"""
    messages: Annotated[Sequence[BaseMessage], operator.add]   # 累加消息历史
    next: str                                                  # Supervisor 决定下一个执行的 Agent
    task_plan: List[str]                                       # 任务分解计划
    resume_context: dict                                       # 简历提取的信息
    final_answer: str                                          # 最终输出
````

````python
# ==================== 5. 定义多个 Specialized Agents （已修复 Prompt） ====================

from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import SystemMessage
from langchain.agents import create_agent

# Researcher Agent
researcher_system_prompt = """你是一位专业的研究助手。擅长从用户简历和互联网中查找信息。
你的任务是深入研究用户的技术背景、项目经验，并收集相关最新技术趋势。
回答要准确、结构化，并提供关键事实。"""

researcher_agent = create_agent(
    model=llm,
    tools=[retriever.as_tool(name="search_resume", description="在用户简历中搜索信息")],
    system_prompt=researcher_system_prompt   # 必须是 str 或 SystemMessage
)

# Coder Agent
coder_system_prompt = """你是一位资深代码工程师（10年 Python + C# 经验）。
擅长生成高质量、可维护的代码，并进行代码审查。
生成代码时要包含详细注释，考虑生产级最佳实践（如错误处理、类型提示、可观测性）。"""

coder_agent = create_agent(
    model=llm,
    tools=[],
    system_prompt=coder_system_prompt
)

# Reviewer / Critic Agent
reviewer_system_prompt = """你是一位严格的代码审查者和质量把关者。
你的任务是仔细审查输出，指出潜在问题、逻辑错误、hallucination 和改进空间。
给出具体、建设性的改进建议，风格专业、严谨且客观。"""

reviewer_agent = create_agent(
    model=llm,
    tools=[],
    system_prompt=reviewer_system_prompt
)


print("✅ 三个 Agent 的 Prompt 已修复并创建完成")
````

````python
# ==================== 6. 创建 Supervisor Agent（路由器） ====================

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from pydantic import BaseModel, Field

class RouteResponse(BaseModel):
    next: str = Field(..., description="下一个要执行的 Agent 名称")

supervisor_prompt = ChatPromptTemplate.from_template("""
你是一个 Multi-Agent 系统的 Supervisor（协调者）。
当前任务是帮助用户进行 AI 职业发展支持（简历优化、面试准备、技术研究等）。

已有对话历史：
{messages}

可用的 Agent 角色：
- Researcher：负责从简历和网络中收集信息
- Coder：负责生成和优化代码
- Reviewer：负责审查输出质量，提出改进意见
- FINISH：当任务已完成时选择此项

请决定下一步应该由哪个 Agent 执行。
如果任务已经可以给出最终答案，请回复 FINISH。

只返回以下格式（不要解释）：
{{"next": "Researcher"}} 或 {{"next": "Coder"}} 或 {{"next": "Reviewer"}} 或 {{"next": "FINISH"}}
""")

supervisor_chain = (
    supervisor_prompt 
    | llm.with_structured_output(RouteResponse)
)

print("✅ Supervisor Agent（路由器）创建完成")
````

````python
# ==================== 7. 构建 LangGraph Multi-Agent 工作流 ====================

from langgraph.graph import StateGraph, END
from langgraph.types import Send

# 定义节点函数
def researcher_node(state: AgentState):
    result = researcher_agent.invoke(state)
    return {"messages": result["messages"]}

def coder_node(state: AgentState):
    result = coder_agent.invoke(state)
    return {"messages": result["messages"]}

def reviewer_node(state: AgentState):
    result = reviewer_agent.invoke(state)
    return {"messages": result["messages"]}

def supervisor_node(state: AgentState):
    response = supervisor_chain.invoke({"messages": state["messages"]})
    return {"next": response.next}

# 构建 Graph
workflow = StateGraph(AgentState)

# 添加节点
workflow.add_node("Researcher", researcher_node)
workflow.add_node("Coder", coder_node)
workflow.add_node("Reviewer", reviewer_node)
workflow.add_node("Supervisor", supervisor_node)

# 设置入口
workflow.set_entry_point("Supervisor")

# 添加条件边（Supervisor 决定路由）
def route_next(state: AgentState):
    return state["next"]

workflow.add_conditional_edges(
    "Supervisor",
    route_next,
    {
        "Researcher": "Researcher",
        "Coder": "Coder",
        "Reviewer": "Reviewer",
        "FINISH": END
    }
)

# 添加各 Agent 执行完后回到 Supervisor
for agent in ["Researcher", "Coder", "Reviewer"]:
    workflow.add_edge(agent, "Supervisor")

# 编译 Graph
multi_agent = workflow.compile()

print("🎉 Multi-Agent 系统构建完成！")
````

````python
# ==================== 8. 可视化 Multi-Agent Graph ====================
from IPython.display import Image as IPyImage, display

try:
    display(IPyImage(multi_agent.get_graph().draw_mermaid_png()))
    print("✅ Multi-Agent 工作流图可视化成功")
except Exception as e:
    print("可视化失败（可选）:", e)
````

````python
# ==================== 9. 测试 Multi-Agent 系统 ====================

from langchain_core.messages import HumanMessage
from IPython.display import Markdown, display

config = {"configurable": {"thread_id": "multi_agent_test_001"}}

test_query = "请帮我准备一个针对 AI Engineer 岗位的简历优化方案，并给出技术面试准备建议"

print("🚀 开始 Multi-Agent 协作任务...\n")
print(f"用户任务: {test_query}\n")



result = multi_agent.invoke({
    "messages": [HumanMessage(content=test_query)],
    "task_plan": [],
    "resume_context": {}
}, config=config)

final_answer = result["messages"][-1].content

display(Markdown(f"### 🎯 Multi-Agent 最终输出\n\n{final_answer}"))
````
